Numpy array operations


```python
import numpy as np
```

```python
arr=np.array([20,30,40,50,60])
print("basic slicing",arr[:4])
```

```python
#                   2023  2024  2025
sales=np.array([[0,1000,2000,32000],  #biriyani
                [1,32000,40000,35675], #pizza
                [2,2000,45000,5600]]) #icecream
print("2025 sales of pizza:",sales[1,2])
print("Biriyani all sales",sales[0,1:])
print(" all sales of 2025",sales[:,3])
print("Pizza and icecream of last 2 yr",sales[1:,2:])
```

sorting


```python
arr=np.array([2,5,3,6,5,4])
print("sorting array with datatype",np.sort(arr))
print(sorted(arr))
```

```python
arr=np.array([[2,1,6],
             [7,3,6],
             [5,76,4]])
print(np.sort(arr,axis=1))
```

Filter

```python
nums=np.array([2,5,3,6,5,4])
evennum=nums[nums%2==0]
print(evennum)
cgpa=np.array([3,5.6,7,6.7,5,6.4])
res=cgpa>4.5
print(res)
```

Fancy indexing

```python
#      2020  2021  2022  2023  2024
sales=np.array([20000,40000,45000,50000,56000,65000])
oddyrsales=[1,3,5]
print("Odd yr sales",sales[oddyrsales])
```

Adding

```python
sale2024=np.array([10,20,30,40])
sale2025=np.array([100,200,300,400])
combined=np.concat((sale2024,sale2025))
print(combined)
```

```python
a=np.array([10,20,30,40])
b=np.array([100,200,300,400])
print(a.shape==b.shape)
```

```python
#Join 2d array
mat=np.array([[2,1,6],
            [5,76,4]])
matrix=np.array([[12,11,16],
             
             [15,16,14]])
print("horizontal",np.hstack([mat,matrix]))
print("vertical",np.vstack([mat,matrix]))
```

Adding and removing data


```python
arr=np.array([1,2,3,4,5,6])
print("Append",np.append(arr,54))
print("insert",np.insert(arr,0,90))
print("split",np.split(arr,3))
print("Remove",np.delete(arr,2))
```
