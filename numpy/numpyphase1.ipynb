```python
#zeros
zeros=np.zeros((2,3))
print(zeros)
#ones
ones=np.ones((3,4))
print(ones)
#full
full=np.full((2,3),5)
print("matrix full 7\n",full)
#random
random=np.random.random((2,2))
print(random)
#sequence
seq=np.arange(1,11,2)
print(seq)
```

```python
vector=np.array([1,2,3,4,5])
print(vector)
matrix=np.array([[1,2,3],[4,5,6]])
print(matrix)
tensor=np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
print(tensor)
```

###Array properties

```python
print(f"Shape:vector={vector.shape}, matrix={matrix.shape}, Tensor={tensor.shape}")
print(f"Dimension:vector={vector.ndim}, matrix={matrix.ndim}, Tensor={tensor.ndim}")
print(f"Size:vector={vector.size}, matrix={matrix.size}, Tensor={tensor.size}")
print(f"Dimension:vector={vector.dtype}, matrix={matrix.dtype}, Tensor={tensor.dtype}")
```
