Advance operations with business example

```python
import numpy as np
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])
# 1. Shape of sales data
print("Shape:", sales_data.shape)


# 2. Sample data for first 3 restaurants
print("First 3 restaurants:")
print(sales_data[:3])


# 3. Total sale every year from all restaurants
print("Total sales every year:")
print(np.sum(sales_data[:, 1:], axis=0))


# 4. Minimum sale per restaurant
print("Minimum sale per restaurant:")
print(np.min(sales_data[:, 1:], axis=1))


# 5. Maximum sale per year
print("Maximum sale per year:")
print(np.max(sales_data[:, 1:], axis=0))


# 6. Average sales per restaurant
print("Average sales per restaurant:")
print(np.mean(sales_data[:, 1:], axis=1))


# 7. Average cumulative sales across all restaurants
print("Cumulative sales:")
print(np.cumsum(sales_data[:, 1:], axis=1))


# 8. Monthly average every year for all restaurants
print("Monthly average every year:")
print(np.mean(sales_data[:, 1:], axis=0) / 12)
```
