Numpy functions


```python
import numpy as np
```

Aggregate functions

```python
arr=np.array([10,20,30,40,50])
print("Sum:",np.sum(arr))
print("Max:",np.max(arr))
print("Min:",np.min(arr))
print("Cumulative Sum:",np.cumsum(arr))
print("Cumulative product:",np.cumprod(arr))
```

statistical functions

```python
import statistics as st
m1=np.array([12,20,34,32,14,23,21])
m2=np.array([16,12,13,18,24,22,32])
print("Mean M1",np.mean(m1),"Mean m2",np.mean(m2))
print("Median M1",np.median(m1),"Median m2",np.median(m2))
print("Mode M1",st.mode(m1),"Mode m2",st.mode(m2))
print("Standard deviation m1",np.std(m1),"Standard deviation m2",np.std(m2))
```
