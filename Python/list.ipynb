```python
#List Methods
colors=["Red","Blue","black","pink"]
print(len(colors))
colors.append("green")
print("after append",colors)
colors.insert(1,"yellow")
print("after insert",colors)
colors.remove("black")
print("after remove",colors)
colors.pop()
print("after pop",colors)
colors[3]="babypink" #replace
print(colors)
```
