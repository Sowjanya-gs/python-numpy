```python
#indexing
str="DARKCHOCOLATE"
#+ve indexing
print(str[4])
print(str[len(str)-1])
#-ve indexing
print(str[-1])
print(str[-len(str)])
```

```python
str="DARKCHOCOLATE"
print(str[0:4])
print(str[4:9])
print(str[4:len(str)])
print(str[9:])
print(str[3::2])
print(str[len(str)-1:3:-1])
```

```python
s="meoww meow meoww"
print(len(s))#find length
print(s.capitalize())
print(s+"Cat") #concat
print(s.replace("meow","meowing"))
print(s.find("e")) #first occurance index
print(s.count("i")) #frequency of substring
print(s.upper()) #uppercase
print(s.lower()) #lowercase
print("is" in s) #check substring is present in string
```
